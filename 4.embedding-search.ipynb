```python
query = "How to reduce heart disease risk?"
```

```python
documents = [
    "Eating fiber reduces heart risk.",
    "Fruits and vegetables lower cardiovascular disease chances.",
    "Buying a new car improves driving comfort.",
    "Regular exercise improves heart health."
]
```

# Semantic(Similarity) Search

1. Dot Product
2. Cosine
3. Euclidean Distance(L2)

```python
import numpy as np
```

```python
def dot_product(a, b):
    return np.dot(a, b)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
def euclidean_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.linalg.norm(a - b)
```

```python
from langchain_openai import OpenAIEmbeddings

openai_embedding_model = OpenAIEmbeddings(model = "text-embedding-3-small")
```

```python
query_embedding = openai_embedding_model.embed_query(query)
```

```python
doc_embeddings = openai_embedding_model.embed_documents(documents)
```

```python
results = []

for doc, doc_emb in zip(documents, doc_embeddings):
    results.append({
        "query": query,
        "document": doc,
        "cosine": cosine_similarity(query_embedding, doc_emb),
        "dot_product": dot_product(query_embedding, doc_emb),
        "euclidean": euclidean_distance(query_embedding, doc_emb)
    })
```

```python
results
```

# Keyword Search

```python
from collections import Counter
```

```python
query_words = query.lower().replace("?", "").split()
```

```python
query_words
```

```python
def keyword_search(documents):
    results = []

    for doc in documents:
        doc_words = doc.lower().replace(".", "").split()
        doc_count = Counter(doc_words)

        print ("doc_count:", doc_count)

        score = sum(doc_count[word] for word in query_words)

        print ("score:", score)

        results.append((doc, score))

    return sorted(results, key=lambda x: x[1], reverse=True)
```

```python
keyword_results = keyword_search(documents)
```
