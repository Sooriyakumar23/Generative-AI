```python
import os
import numpy as np
import pandas as pd
import gensim
import nltk
```

```python
nltk.download('punkt')           # sentence & word tokenizer
nltk.download('stopwords')       # common words to filter out (the, is, at...)
nltk.download('wordnet')         # word lemmatization
nltk.download('punkt_tab')       # needed in newer nltk versions (3.8+)
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
for file in os.listdir('./word2vec_training_data/'):
    print(file)
```

```python
folder_path = r".\word2vec_training_data"

story = []

for filename in os.listdir(folder_path):
  file_path=os.path.join(folder_path,filename)

  with open(file_path, encoding='unicode_escape') as f:
    corpus=f.read()

  raw_sent=sent_tokenize(corpus)
  
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
len(story)
```

```python
story
```

Word2Vec Model Comparison

| Feature | CBOW | Skip-gram |
|---------|------|-----------|
| Speed | Fast | Slow |
| Accuracy | Medium | High |
| Rare words | Weak | Strong |
| Data requirement | Less | More |

Architecture Comparison

| Part | CBOW | Skip-gram |
|------|------|-----------|
| Input vectors | Multiple | Single |
| Hidden layer | Average | Direct |
| Output | Single | Multiple |

```python
# Custom Word2Vec model initialized for training

custom_model =gensim.models.Word2Vec(
    window=10,
    min_count=1,
    vector_size=150 
)
```

```python
custom_model.build_vocab(story)
```

```python
custom_model.corpus_count
```

```python
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=10)
```
