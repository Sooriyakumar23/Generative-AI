# 1. Hugging Face

Open source models

```python
from sentence_transformers import SentenceTransformer
```

```python
word = "machine"
```

```python
model_st = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
word_embedding = model_st.encode(word)
```

```python
word_embedding
```

```python
word_embedding.shape
```

```python
sentence = "The cat is on the table."
```

```python
sentence_embedding = model_st.encode(sentence)
```

```python
sentence_embedding.shape
```

```python
paragraph = """
The cat is on the table. The dog is in the garden.
The sun is shining. The sky is blue.
The weather is nice. The birds are singing.
The flowers are blooming. The grass is green.
The water is clear. The air is fresh.
"""
```

# 3. Image Embedding

```python
from transformers import CLIPModel, CLIPProcessor
```

```python
from PIL import Image
```

```python
image = Image.open("C:\\PERSONAL\\FullStackGenAI-KNAcademy\\Sooriyakumar\\image\\image1.jpg")
```

```python
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
```

```python
inputs = processor(images=image, return_tensors="pt")
```

```python
inputs
```

```python
import torch
```

```python
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
```

```python
# Get embedding

with torch.no_grad():
    image_embedding = model.get_image_features(**inputs)
```
